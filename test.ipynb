```python
import pickle
import networkx as nx
import json

# 1. Load the graph
file_path = '/Users/kerem/Lecture_Materials/CS455/Project/SuSchedule-r/data/degree_graphs/BSCS-DM/202201/Core_Elective.gpickle'
with open(file_path, 'rb') as f:
    G = pickle.load(f)

# 2. Access the data for a specific node
node_id = "IF 100"

if G.has_node(node_id):
    # G.nodes[node_id] returns the dictionary of attributes for that node
    node_data = G.nodes[node_id]

    # 3. Print it in a highly readable, indented JSON format
    print(f"Data for {node_id}:")
    print(json.dumps(node_data, indent=2))

    # --- How to extract specific pieces of this data ---

    # Get the title
    title = node_data.get("title", "No title found")
    print(f"\nCourse Title: {title}")

    # Get the prerequisites list (navigating the nested dictionary)
    if "prereq_expr" in node_data and "operands" in node_data["prereq_expr"]:
        print("\nPrerequisites needed:")
        for prereq in node_data["prereq_expr"]["operands"]:
            print(f"- {prereq['course']} (Min Grade: {prereq['min_grade']})")

else:
    print(f"Node '{node_id}' does not exist in the graph.")
```
