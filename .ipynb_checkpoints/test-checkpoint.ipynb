```python
import pickle
import networkx as nx
from pyvis.network import Network


# 1. Load the file
with open('/Users/kerem/Lecture_Materials/CS455/Project/SuSchedule-r/data/degree_graphs/BSCS-DM/Core_Elective.gpickle', 'rb') as f:
    G = pickle.load(f)

# 2. Create a PyVis network canvas
# (Set directed=True if your graph is directed, which looks best for trees)
net = Network(height="800px", width="100%", bgcolor="#ffffff", font_color="black", directed=True)

# 3. Load your NetworkX graph into PyVis
net.from_nx(G)

# 4. Inject settings to force a clean, top-down Tree/Hierarchical layout
net.set_options("""
var options = {
  "layout": {
    "hierarchical": {
      "enabled": true,
      "direction": "UD",
      "sortMethod": "directed",
      "nodeSpacing": 150,
      "treeSpacing": 200
    }
  },
  "physics": {
    "hierarchicalRepulsion": {
      "nodeDistance": 150
    },
    "solver": "hierarchicalRepulsion"
  }
}
""")

# 5. Generate and open the interactive webpage
net.show("my_tree_graph.html", notebook=False)
```
